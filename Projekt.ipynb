```python
import pandas as pd
import numpy as np
import missingno as msno
from skimpy import skim
```

```python
# Projekt 
gym_org = pd.read_csv('./Data/silownia_new.csv')
gym = gym_org.copy() # Let's create a copy of the original data to work on it and keep the original one as a reference
#gym.info()
#print(gym.isna().any())
#msno.matrix(gym)  # It's abe to create it but... it's so unreadable 
skim(gym) #This rocks! Can replace all previous calls (info(), isna().any() and even msno.matrix )
```

```python
#Let's see the lacks distribution
import seaborn as sns
sns.heatmap(gym.isna());
```

```python
#Quick replace Gender: string to int
gender_map = {'Male': 1, 'Female': 0}
gym['Gender'] = gym['Gender'].map(gender_map)
```

```python
workout_type_map = {'HIIT':0, 'Strength':1, 'Cardio':2, 'Yoga':3}
gym['Workout_Type'] = gym['Workout_Type'].map(workout_type_map)
```

```python
#We can read those missing values step by step - manually
age = gym['Age'].copy()
print(f'Amount of missing values(Age): {age.isna().sum()}')

workout = gym['Workout_Type'].copy()
print(f'Amount of missing values(Workout_type): {workout.isna().sum()}')

bmi = gym['BMI'].copy()
print(f'Amount of missing values(BMI): {bmi.isna().sum()}')
```

```python
#BMI = Weight / Height**2
#Weight and Height doesn't lack values so we can just evaluate them

gym_bmi = gym.copy()
gym_bmi['EVAL_BMI'] = gym_bmi['Weight (kg)'] / (gym_bmi['Height (m)'] ** 2)

non_nan_bmi = gym_bmi.loc[gym_bmi['BMI'] != np.nan]

gym_bmi['EVAL_BMI_diff'] = (gym_bmi['EVAL_BMI'] - gym_bmi['BMI'])
gym_bmi[['EVAL_BMI', 'BMI']].plot.hist(bins=50, alpha=0.5)
```

W danych wyliczone BMI wg:

$BMI = \frac{masa\ (kg)}{wzrost^2\ (m)}$

nie pokrywa się całkowicie z wartością podaną w danych. Oryginalne wartości zostawiamy, brakujace wypełniamy wg. wzoru

```python
gym['BMI'] = gym['BMI'].fillna(gym['Weight (kg)'] / (gym['Height (m)'] ** 2))
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.impute import KNNImputer

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

numeric_cols = gym.select_dtypes(include=np.number).columns
gym_numeric = gym.copy()[numeric_cols]

original_age = gym_numeric['Age'].copy()
missing_mask_age = original_age.isna()
ground_truth_age = original_age.fillna(original_age.median())
```

```python
errors = []
neighbors = range(2, 40)

for k in neighbors:
    imputer = KNNImputer(n_neighbors=k)
    imputed = imputer.fit_transform(gym_numeric)
    imputed_df = pd.DataFrame(imputed, columns=numeric_cols, index=gym_numeric.index)
    
    if missing_mask_age.any():
        imputed_age_at_missing = imputed_df.loc[missing_mask_age, 'Age']
        ground_truth_at_missing = ground_truth_age.loc[missing_mask_age]
        mse = mean_squared_error(ground_truth_at_missing, imputed_age_at_missing)
        errors.append(mse)
    else:
        errors.append(0)

plt.plot(neighbors, errors, marker='o')
plt.xlabel('Liczba sąsiadów (k)')
plt.ylabel('MSE')
plt.title('Elbow method dla KNNImputer')
plt.show()
```

Dla wieku (Age) mniej więcej od wartości 7 następuje przegięcie - dobre wyniki powinniśmy juz dostać wyciagając średnią dla 17 sąsiadów.

```python
imputer = KNNImputer(n_neighbors=17) 
imputed = imputer.fit_transform(gym_numeric)
imputed_df = pd.DataFrame(imputed, columns=numeric_cols, index=gym_numeric.index)
```

```python

#Finally replace the org values with imputed
gym['Age'] = imputed_df['Age']
gym['Workout_Type'] = imputed_df['Workout_Type']
```

Na koniec odwracamy na powrót wartości zmiennej kategorycznej

```python
reverse_map = {v: k for k, v in workout_type_map.items()}
gym['Workout_Type'] = imputed_df['Workout_Type'].round().astype(int).map(reverse_map)
```
